```python
import fitz  # PyMuPDF

pdf_path = r'C:\Users\eenglish\OneDrive - Epic\Desktop\11-22-63.pdf'
output_path = "edited_document.pdf"

# Open the existing PDF
pdf = fitz.open(pdf_path)

# Select the page to edit
page = pdf[0]  # Edit the first page

# Add text at specified location
page.insert_text((100, 100), "Added text using PyMuPDF", fontsize=12)

# Add an image
image_rect = fitz.Rect(100, 150, 200, 250)
page.insert_image(image_rect, filename="image_to_add.png")

# Save the edited PDF
pdf.save(output_path)
pdf.close()
```
