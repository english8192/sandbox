```python
import re

def find_words_with_vowels(filename):
    # Define a regular expression for three or more consecutive vowels
    vowels_pattern = re.compile(r"[aeiouAEIOU]{4,}")
   # vowels_pattern = re.compile(r"ology")
    # List to store words with three or more consecutive vowels
    matching_words = []
    
    try:
        with open(filename, 'r') as file:
            for line in file:
                word = line.strip()  # Remove any extra spaces or newlines
                if vowels_pattern.search(word):  # Check if the word matches the pattern
                    matching_words.append(word)
    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []

    return matching_words

# Example usage:
input_file = "words_alpha.txt"  # Replace with your file name
results = find_words_with_vowels(input_file)

if results:
    print("Words with three or more consecutive vowels:")
    for word in results:
        print(word)
else:
    print("No words with three or more consecutive vowels were found.")
```
